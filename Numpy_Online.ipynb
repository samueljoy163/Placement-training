**NUMPY**

What is NumPy?

A library for handling arrays (like lists but more powerful).

Provides support for mathematical operations on large datasets efficiently.

Backbone of data science, ML, AI, and scientific computing in Python.

 Why NumPy?

✅ Faster than Python lists

✅ Uses less memory

✅ Provides many built-in mathematical functions

✅ Works well with Pandas, SciPy, Scikit-Learn, TensorFlow etc.

Creating Arrays

Arrays are very important in data science and programming because they provide a structured and efficient way to store and manipulate data

Arithmetic Operations

```python
# Element-wise operations
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

sum_arr = arr1 + arr2                      # Addition
sub_arr = arr1 - arr2                      # Subtraction
prod_arr = arr1 * arr2                     # Multiplication
div_arr = arr1 / arr2                      # Division
exp_arr = np.exp(arr1)                     # Exponential
log_arr = np.log(arr2)                     # Logarithm

print(sum_arr)

print(log_arr)
```

matrix multiplication

```python
A = np.array([[1,2],[3,4]])
B = np.array([[5,6],[7,8]])

print(A.dot(B))      # matrix multiplication
print(np.matmul(A,B))
```

```python
array_2d = np.array([[1, 2], [3, 4]])
print(array_2d)
print(array_2d[0, 0])                      # Element at row 0, column 1
print(array_2d[:, 1])                      # All rows, column 1

array_2d[0,1]
array_2d[1,1]
print(array_2d)
```

**MATPLOTLIB**

```python
import matplotlib.pyplot as plt
import numpy as np
```

Simple Line Plot

```python
x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 6, 8, 10])

plt.plot(x, y)
plt.title("line graph")
plt.xlabel("data")
plt.ylabel("frequency")
plt.grid(True)
# line graph
plt.show()
```

Adding Labels & Title

```python
plt.plot(x, y,label="Data")
plt.xlabel("height if students")
plt.ylabel("Y-axis")
plt.title("Simple Line Graph")
plt.legend()
plt.grid(True)
plt.show()

```

Line Styles, Colors & Markers

```python
plt.plot(x, y, color='blue', linestyle='--', marker='*')
plt.show()
```

Common options:

Colors: 'red', 'blue', 'green'

Linestyle: '-', '--', ':', '-.'

Marker: 'o', '*', 's', 'd', 'x'

Scatter Plot

```python
plt.scatter(x, y, color='purple')
plt.show()
plt.grid()
```

Bar Chart

```python
categories = ['A', 'B', 'C', 'D']
values = [10, 20, 15, 30]

plt.bar(categories, values, color='orange')
plt.show()
```

Histogram

```python
data = np.random.randn(1000)  # 1000 random numbers
plt.hist(data, bins=20, color='green', edgecolor='black')
plt.show()
```

Pie Chart

```python
sizes = [20, 30, 25, 25]
labels = ['A', 'B', 'C', 'D']

plt.pie(sizes, labels=labels, autopct='%1.2f%%', startangle=90)
plt.show()
```

Subplots (Multiple Plots in One Figure)

```python
plt.subplot(1, 2, 1)   # 1 row, 2 columns, 1st plot
plt.plot(x, y)

plt.subplot(1, 2, 2)   # 2nd plot
plt.bar(categories, values)

plt.show()
```

Figure Size & Grid

```python
plt.figure(figsize=(6,4))   # width, height
plt.plot(x, y)
plt.grid(True)              # show grid
plt.show()
```

Save a Plot

```python
x = np.linspace(0, 5, 50)
y = np.exp(x)

plt.plot(x, y)
plt.title("Exponential Growth")
plt.savefig("exponential_plot.png")  # save as PNG
plt.show()
```
