### Cleaning Text Fields

```python
for col in df.select_dtypes(include=['object']).columns:
    # Remove leading/trailing spaces
    df[col] = df[col].str.strip()
    # Replace multiple spaces with a single space
    df[col] = df[col].str.replace(r'\s+', ' ', regex=True)

print("Text fields cleaned successfully. Displaying a sample of the 'Restaurant Name' column after cleaning:")
display(df['Restaurant Name'].head())
```

```python
df['Restaurant Name'] = df['Restaurant Name'].str.title()
df['City'] = df['City'].str.title()
df['Locality'] = df['Locality'].str.title()
df['Cuisines'] = df['Cuisines'].str.title()

print("Text columns standardized to title case.")
display(df[['Restaurant Name', 'City', 'Locality', 'Cuisines']].head())
```

### Encoding Binary Columns

```python
binary_map = {'Yes': 1, 'No': 0}

columns_to_encode = [
    'Has Table booking',
    'Has Online delivery',
    'Is delivering now',
    'Switch to order menu'
]

for col in columns_to_encode:
    if col in df.columns:
        df[col] = df[col].map(binary_map)
        print(f"Column '{col}' encoded successfully.")
    else:
        print(f"Column '{col}' not found in DataFrame.")

print("\nDisplaying the first 5 rows of the encoded columns:")
display(df[columns_to_encode].head())
```

### One-Hot Encoding for Categorical Columns

```python
columns_to_onehot = ['Currency', 'Rating text']

for col in columns_to_onehot:
    if col in df.columns:
        dummies = pd.get_dummies(df[col], prefix=col, drop_first=True)
        df = pd.concat([df, dummies], axis=1)
        df.drop(col, axis=1, inplace=True)
        print(f"Column '{col}' one-hot encoded and original column dropped.")
    else:
        print(f"Column '{col}' not found in DataFrame.")

print("\nDataFrame head after one-hot encoding:")
display(df.head())
```

### Outlier Detection using IQR

```python
numerical_cols_for_outliers = ['Average Cost for two', 'Votes', 'Aggregate rating']

for col in numerical_cols_for_outliers:
    if col in df.columns:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
        print(f"Column '{col}': {len(outliers)} outliers detected.")
    else:
        print(f"Column '{col}' not found in DataFrame.")
```

### Outlier Removal and Dataset Size Comparison

Removing outliers can help improve the performance of machine learning models by reducing noise and extreme values that might skew the data distribution. I will remove outliers from 'Average Cost for two', 'Votes', and 'Aggregate rating' using the IQR method and then compare the DataFrame's size before and after removal.

```python
print(f"DataFrame shape before outlier removal: {df.shape}")

def remove_outliers_iqr(df_input, col):
    Q1 = df_input[col].quantile(0.25)
    Q3 = df_input[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    df_filtered = df_input[(df_input[col] >= lower_bound) & (df_input[col] <= upper_bound)]
    return df_filtered

# Columns for outlier removal
numerical_cols_for_outliers = ['Average Cost for two', 'Votes', 'Aggregate rating']

for col in numerical_cols_for_outliers:
    initial_rows = len(df)
    df = remove_outliers_iqr(df, col)
    rows_removed = initial_rows - len(df)
    print(f"Outliers removed from '{col}': {rows_removed} rows. New shape: {df.shape}")

print(f"\nDataFrame shape after outlier removal: {df.shape}")
```
